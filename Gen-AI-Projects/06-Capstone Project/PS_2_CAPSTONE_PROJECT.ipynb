```python
print("99 Akanksha Netkar")

import sqlite3
import pandas as pd

DB_NAME = "railway_database.db"


def get_database_connection():
    return sqlite3.connect(DB_NAME)


class TicketAgent:

    def run(self, ticket_id):

        connection = get_database_connection()

        query = """
        SELECT *
        FROM tickets
        WHERE Ticket_ID = ?
        """

        result = pd.read_sql_query(
            query,
            connection,
            params=(ticket_id,)
        )

        connection.close()

        if result.empty:
            return {
                "status": "error",
                "message": f"Ticket ID {ticket_id} was not found."
            }

        return {
            "status": "success",
            "data": result.iloc[0].to_dict()
        }


class TransactionAgent:

    def run(self, transaction_id):

        connection = get_database_connection()

        query = """
        SELECT *
        FROM tickets
        WHERE Transaction_ID = ?
        """

        result = pd.read_sql_query(
            query,
            connection,
            params=(transaction_id,)
        )

        connection.close()

        if result.empty:
            return {
                "status": "error",
                "message": f"Transaction ID {transaction_id} was not found."
            }

        return {
            "status": "success",
            "data": result.iloc[0].to_dict()
        }


class PassengerAgent:

    def run(self, passenger_name):

        connection = get_database_connection()

        query = """
        SELECT *
        FROM tickets
        WHERE Passenger_Name LIKE ?
        """

        result = pd.read_sql_query(
            query,
            connection,
            params=(f"%{passenger_name}%",)
        )

        connection.close()

        if result.empty:
            return {
                "status": "error",
                "message":
                    f"No passenger found matching {passenger_name}."
            }

        return {
            "status": "success",
            "records": result.to_dict(
                orient="records"
            )
        }


class AdminAgent:

    def run(self, query):

        connection = get_database_connection()

        result = pd.read_sql_query(
            "SELECT * FROM tickets",
            connection
        )

        connection.close()

        return {
            "status": "success",
            "message": "Administrative information retrieved.",
            "total_tickets": len(result),
            "confirmed": int(
                (result["Status"] == "Confirmed").sum()
            ),
            "waiting": int(
                (result["Status"] == "Waiting List").sum()
            ),
            "cancelled": int(
                (result["Status"] == "Cancelled").sum()
            )
        }


ticket_agent = TicketAgent()
transaction_agent = TransactionAgent()
passenger_agent = PassengerAgent()
admin_agent = AdminAgent()

print("✅ All railway agents created!")
```

```python
print("99 Akanksha Netkar")

import re

SUPERVISOR_PROMPT = """
You are the Supervisor Agent of an Indian Railways
Administrator System.

Choose exactly ONE agent.

TICKET_AGENT:
Use for Ticket ID, ticket details, ticket status,
train or journey information.

TRANSACTION_AGENT:
Use for Transaction ID or finding a Ticket ID
using a Transaction ID.

PASSENGER_AGENT:
Use for passenger name and passenger booking information.

ADMIN_AGENT:
Use for statistics, reports, summaries and
general administrative requests.

Return ONLY one of:

TICKET_AGENT
TRANSACTION_AGENT
PASSENGER_AGENT
ADMIN_AGENT

Do not explain your answer.
"""


def qwen_supervisor(query):

    messages = [
        {
            "role": "system",
            "content": SUPERVISOR_PROMPT
        },
        {
            "role": "user",
            "content": query
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():

        outputs = model.generate(
            **inputs,
            max_new_tokens=5,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    response = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    response = response.strip().upper()

    if "TRANSACTION_AGENT" in response:
        return "TRANSACTION_AGENT"

    if "PASSENGER_AGENT" in response:
        return "PASSENGER_AGENT"

    if "TICKET_AGENT" in response:
        return "TICKET_AGENT"

    return "ADMIN_AGENT"


print("✅ Qwen Supervisor created!")
```

```python
print("99 Akanksha Netkar")

from typing import TypedDict
from langgraph.graph import StateGraph, START, END


class RailwayState(TypedDict):

    user_query: str
    agent: str
    database_result: dict
    final_response: str


def supervisor_node(state):

    query = state["user_query"]

    decision = qwen_supervisor(query)

    mapping = {
        "TICKET_AGENT": "ticket",
        "TRANSACTION_AGENT": "transaction",
        "PASSENGER_AGENT": "passenger",
        "ADMIN_AGENT": "admin"
    }

    selected_agent = mapping.get(
        decision,
        "admin"
    )

    return {
        "agent": selected_agent
    }


def ticket_node(state):

    query = state["user_query"]

    match = re.search(
        r"TKT\d+",
        query.upper()
    )

    if not match:

        return {
            "database_result": {
                "status": "error",
                "message": "No Ticket ID found."
            }
        }

    ticket_id = match.group()

    return {
        "database_result":
            ticket_agent.run(ticket_id)
    }


def transaction_node(state):

    query = state["user_query"]

    match = re.search(
        r"TXN\d+",
        query.upper()
    )

    if not match:

        return {
            "database_result": {
                "status": "error",
                "message": "No Transaction ID found."
            }
        }

    transaction_id = match.group()

    return {
        "database_result":
            transaction_agent.run(transaction_id)
    }


def passenger_node(state):

    query = state["user_query"]

    connection = get_database_connection()

    all_passengers = pd.read_sql_query(
        "SELECT Passenger_Name FROM tickets",
        connection
    )

    connection.close()

    query_lower = query.lower()

    found_name = None

    for name in all_passengers["Passenger_Name"]:

        if name.lower() in query_lower:
            found_name = name
            break

        first_name = name.split()[0]

        if first_name.lower() in query_lower:
            found_name = first_name
            break

    if found_name is None:

        return {
            "database_result": {
                "status": "error",
                "message":
                    "Passenger name could not be identified."
            }
        }

    return {
        "database_result":
            passenger_agent.run(found_name)
    }


def admin_node(state):

    return {
        "database_result":
            admin_agent.run(
                state["user_query"]
            )
    }


def final_response_node(state):

    data = state["database_result"]

    return {
        "final_response": str(data)
    }


def route_from_supervisor(state):

    return state["agent"]


graph_builder = StateGraph(
    RailwayState
)

graph_builder.add_node(
    "supervisor",
    supervisor_node
)

graph_builder.add_node(
    "ticket",
    ticket_node
)

graph_builder.add_node(
    "transaction",
    transaction_node
)

graph_builder.add_node(
    "passenger",
    passenger_node
)

graph_builder.add_node(
    "admin",
    admin_node
)

graph_builder.add_node(
    "final_response",
    final_response_node
)


graph_builder.add_edge(
    START,
    "supervisor"
)


graph_builder.add_conditional_edges(
    "supervisor",
    route_from_supervisor,
    {
        "ticket": "ticket",
        "transaction": "transaction",
        "passenger": "passenger",
        "admin": "admin"
    }
)


graph_builder.add_edge(
    "ticket",
    "final_response"
)

graph_builder.add_edge(
    "transaction",
    "final_response"
)

graph_builder.add_edge(
    "passenger",
    "final_response"
)

graph_builder.add_edge(
    "admin",
    "final_response"
)

graph_builder.add_edge(
    "final_response",
    END
)


railway_graph = graph_builder.compile()

print("✅ LangGraph architecture created!")
```

```python
print("99 Akanksha Netkar")

import requests

try:

    response = requests.get(
        "http://127.0.0.1:8501",
        timeout=10
    )

    print("Status code:", response.status_code)

    if response.status_code == 200:

        print("✅ STREAMLIT IS RUNNING!")
        print("✅ Dashboard is ready.")

    else:

        print(
            "❌ Streamlit returned:",
            response.status_code
        )

except Exception as e:

    print("❌ Streamlit is not running.")
    print("Error:", e)


print("\n===== STREAMLIT LOG =====")

if os.path.exists("streamlit.log"):

    with open(
        "streamlit.log",
        "r",
        encoding="utf-8",
        errors="ignore"
    ) as f:

        print(
            f.read()[-4000:]
        )
```
